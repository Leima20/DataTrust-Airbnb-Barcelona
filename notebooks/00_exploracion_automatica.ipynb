Este notebook **no forma parte de la serie numerada 01-06** de hallazgos del
proyecto. Es un paso previo de comprensión: generar un gráfico por cada
columna de las 3 tablas crudas (`listings`, `reviews`, `calendar`), eligiendo
automáticamente el tipo de gráfico según el tipo de dato (histograma/boxplot

No se excluye ninguna columna, aunque algunas (texto libre como `comments`,

## 1. Preparación: librerías y funciones

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from graficos_utils import tipo_de_columna, graficar_numerica, graficar_categorica, graficar_fecha, explorar_dataset

pd.set_option("display.max_columns", None)
```

## 2. Tabla `listings`

Se cargan las columnas de fecha explícitamente (`last_scraped`, `calendar_last_scraped`,
`host_since`, `first_review`, `last_review`) para que `tipo_de_columna` las
reconozca como `'fecha'` en vez de texto — si no, saldrían como gráfico de
barras de valores en vez de gráfico de años.

```python
COLUMNAS_FECHA_LISTINGS = ["last_scraped", "calendar_last_scraped", "host_since", "first_review", "last_review"]

listings = pd.read_csv("../data/airbnb_bcn/listings.csv", low_memory=False, parse_dates=COLUMNAS_FECHA_LISTINGS)
print("listings:", listings.shape)

rutas_listings = explorar_dataset(listings, carpeta_salida="../graficos_exploracion/listings")
print(f"Gráficos generados para listings: {len(rutas_listings)}")
```

## 3. Tabla `reviews`

```python
reviews = pd.read_csv("../data/airbnb_bcn/reviews.csv", low_memory=False, parse_dates=["date"])
print("reviews:", reviews.shape)

rutas_reviews = explorar_dataset(reviews, carpeta_salida="../graficos_exploracion/reviews")
print(f"Gráficos generados para reviews: {len(rutas_reviews)}")
```

## 4. Tabla `calendar`

La más grande (más de 5,6 millones de filas) — puede tardar algo más que las otras dos.

```python
calendar = pd.read_csv("../data/airbnb_bcn/calendar.csv", low_memory=False, parse_dates=["date"])
print("calendar:", calendar.shape)

rutas_calendar = explorar_dataset(calendar, carpeta_salida="../graficos_exploracion/calendar")
print(f"Gráficos generados para calendar: {len(rutas_calendar)}")
```

## 5. Resumen
